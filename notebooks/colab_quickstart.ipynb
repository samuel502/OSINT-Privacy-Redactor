# Privacy Footprint Agent - Colab quickstart
Add `GROQ_API_KEY` (and/or `GEMINI_API_KEY`) in Colab **Secrets** (key icon, left sidebar) and enable notebook access.

The UI opens inside this notebook. A public share link is optional and needs a login: add the Secrets `GRADIO_SHARE` = `1`, `GRADIO_USERNAME` and `GRADIO_PASSWORD` (12+ characters). Without both login values the app refuses to start a public link.
